# Create a subset training data of 100 classes each containing 50 images

```python
import os
import shutil
import random

full_dataset_path = "/content/dataset_folder/University-Release/train/"
subset_path = "/content/subset_dataset"
subset_size_per_class = 50
num_classes_to_sample = 100

modalities = ['drone', 'satellite']

sat_classes = set(os.listdir(os.path.join(full_dataset_path, 'satellite')))
drone_classes = set(os.listdir(os.path.join(full_dataset_path, 'drone')))
shared_classes = list(sat_classes & drone_classes)

selected_classes = random.sample(shared_classes, min(num_classes_to_sample, len(shared_classes)))
print(f"Selected {len(selected_classes)} classes.")

for modality in modalities:
    modality_full_path = os.path.join(full_dataset_path, modality)
    modality_subset_path = os.path.join(subset_path, modality)
    os.makedirs(modality_subset_path, exist_ok=True)

    for cls in selected_classes:
        class_full_path = os.path.join(modality_full_path, cls)
        class_subset_path = os.path.join(modality_subset_path, cls)
        os.makedirs(class_subset_path, exist_ok=True)

        all_images = [f for f in os.listdir(class_full_path) if f.lower().endswith(('.jpg','.png','.jpeg'))]
        if not all_images:
            print(f"No images found in {class_full_path}, skipping.")
            continue

        sampled_images = random.sample(all_images, min(subset_size_per_class, len(all_images)))

        for img_name in sampled_images:
            src = os.path.join(class_full_path, img_name)
            dst = os.path.join(class_subset_path, img_name)
            shutil.copyfile(src, dst)

print("Subset folder with randomly selected classes and images created successfully!")
```

# Resize and Save Training Images

```python
import os
from PIL import Image
from pathlib import Path
from tqdm import tqdm

def resize_and_save_dataset(input_root, output_root, size=(224, 224)):
    for mode in ['satellite', 'drone']:
        input_dir = Path(input_root) / mode
        output_dir = Path(output_root) / mode
        output_dir.mkdir(parents=True, exist_ok=True)

        classes = sorted(os.listdir(input_dir))
        for cls in tqdm(classes, desc=f"Resizing {mode} images"):
            input_cls_dir = input_dir / cls
            output_cls_dir = output_dir / cls
            output_cls_dir.mkdir(parents=True, exist_ok=True)

            for img_name in os.listdir(input_cls_dir):
                input_img_path = input_cls_dir / img_name
                output_img_path = output_cls_dir / img_name.replace(".png", ".jpg")

                try:
                    with Image.open(input_img_path) as img:
                        img = img.convert("RGB")
                        img = img.resize(size, Image.BILINEAR)
                        img.save(output_img_path, format="JPEG", quality=90)
                except Exception as e:
                    print(f"Error processing {input_img_path}: {e}")

# Set paths
original_dataset = "/content/subset_dataset"
resized_dataset = "/content/dataset_resized"

# Run it
resize_and_save_dataset(original_dataset, resized_dataset)
```

# Auto Tuning Parameters

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import transforms
from dualresnet import DualResNet
from Modified_Preprocessing import PairedU1652Dataset
from sklearn.metrics import accuracy_score

# --- Search space ---
LEARNING_RATES = [1e-3, 1e-4]
WEIGHT_DECAYS = [0, 1e-5]
DROPOUT_RATES = [0.0, 0.3]

# --- Setup ---
data_dir = "/content/dataset_resized"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
batch_size = 8
num_epochs = 2  # Fast tuning loop

# --- Data transforms ---
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.1, contrast=0.1),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load dataset and use a small subset for speed
dataset = PairedU1652Dataset(root_dir=data_dir, transform_sat=transform, transform_drone=transform)
subset_size = min(len(dataset), 200)
subset, _ = random_split(dataset, [subset_size, len(dataset) - subset_size])
dataloader = DataLoader(subset, batch_size=batch_size, shuffle=True, num_workers=2)

# --- Simple evaluation ---
def evaluate(model, dataloader):
    model.eval()
    preds, targets = [], []
    with torch.no_grad():
        for x1, x2, labels in dataloader:
            x1, x2 = x1.to(device), x2.to(device)
            out1, out2, _, _ = model(x1, x2)
            avg_output = (out1 + out2) / 2
            preds.extend(avg_output.argmax(1).cpu().numpy())
            targets.extend(labels.numpy())
    return accuracy_score(targets, preds)

# --- Auto-tuning loop ---
results = []

for lr in LEARNING_RATES:
    for wd in WEIGHT_DECAYS:
        for dr in DROPOUT_RATES:
            print(f"\nTesting config: lr={lr}, weight_decay={wd}, dropout={dr}")

            model = DualResNet(num_classes=100).to(device)

            # Freeze backbones
            for param in model.backbone1.parameters():
                param.requires_grad = False
            for param in model.backbone2.parameters():
                param.requires_grad = False

            # Add dropout after feature extraction
            model.classifier1 = nn.Sequential(
                nn.Dropout(dr), nn.Linear(512, 100)
            )
            model.classifier2 = nn.Sequential(
                nn.Dropout(dr), nn.Linear(512, 100)
            )

            optimizer = optim.Adam(
                list(model.classifier1.parameters()) + list(model.classifier2.parameters()),
                lr=lr, weight_decay=wd
            )
            criterion = nn.CrossEntropyLoss()

            # --- Training loop ---
            model.train()
            for epoch in range(num_epochs):
                for sat_img, drone_img, labels in dataloader:
                    sat_img, drone_img, labels = sat_img.to(device), drone_img.to(device), labels.to(device)

                    optimizer.zero_grad()
                    out1, out2, _, _ = model(sat_img, drone_img)

                    loss1 = criterion(out1, labels)
                    loss2 = criterion(out2, labels)
                    loss = (loss1 + loss2) / 2
                    loss.backward()
                    optimizer.step()

            acc = evaluate(model, dataloader)
            print(f"Config result: Accuracy = {acc:.4f}")
            results.append((lr, wd, dr, acc))

# --- Report best config ---
best = max(results, key=lambda x: x[3])
print("\nBest Config:")
print(f"LR: {best[0]}, Weight Decay: {best[1]}, Dropout: {best[2]}, Accuracy: {best[3]:.4f}")
```

# Fine Tuning The Model Weights

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import transforms
from dualresnet import DualResNet
from Modified_Preprocessing import PairedU1652Dataset


def fine_tune_classifier():
    # --- Config ---
    model_path = "/content/ACMMM23-Solution-MBEG/weights/best_model.pth"  # pretrained weights
    save_path = "/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth"
    data_dir = "/content/dataset_resized"
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    # Best config from AutoTuning
    lr = 0.001
    weight_decay = 0
    dropout = 0.3
    num_epochs = 15
    batch_size = 8

    # --- Transforms ---
    transform = transforms.Compose([
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=15),
        transforms.ColorJitter(brightness=0.1, contrast=0.1),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ])

    dataset = PairedU1652Dataset(root_dir=data_dir,
                                  transform_sat=transform,
                                  transform_drone=transform)
    dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True, num_workers=2)

    # --- Load model and modify classifier ---
    model = DualResNet(num_classes=100).to(device)
    model.load_state_dict(torch.load(model_path, map_location=device))

    # Freeze backbones
    for param in model.backbone1.parameters():
        param.requires_grad = False
    for param in model.backbone2.parameters():
        param.requires_grad = False

    # Apply dropout in classifier layers
    model.classifier1 = nn.Sequential(nn.Dropout(dropout), nn.Linear(512, 100))
    model.classifier2 = nn.Sequential(nn.Dropout(dropout), nn.Linear(512, 100))

    optimizer = optim.Adam(
        list(model.classifier1.parameters()) + list(model.classifier2.parameters()),
        lr=lr, weight_decay=weight_decay
    )
    criterion = nn.CrossEntropyLoss()

    # --- Training loop ---
    model.train()
    best_acc = 0.0

    for epoch in range(num_epochs):
        total_loss = 0
        correct1 = correct2 = total = 0

        for sat_img, drone_img, labels in dataloader:
            sat_img, drone_img, labels = sat_img.to(device), drone_img.to(device), labels.to(device)

            optimizer.zero_grad()
            out1, out2, _, _ = model(sat_img, drone_img)

            loss1 = criterion(out1, labels)
            loss2 = criterion(out2, labels)
            loss = (loss1 + loss2) / 2
            loss.backward()
            optimizer.step()

            total_loss += loss.item() * labels.size(0)
            correct1 += (out1.argmax(1) == labels).sum().item()
            correct2 += (out2.argmax(1) == labels).sum().item()
            total += labels.size(0)

        acc1 = correct1 / total
        acc2 = correct2 / total
        avg_acc = (acc1 + acc2) / 2
        avg_loss = total_loss / total

        print(f"Epoch [{epoch+1}/{num_epochs}] | Loss: {avg_loss:.4f} | Acc1: {acc1:.4f}, Acc2: {acc2:.4f} | Avg Acc: {avg_acc:.4f}")

        if avg_acc > best_acc:
            best_acc = avg_acc
            torch.save(model.state_dict(), save_path)
            print(f"New best model saved at epoch {epoch+1} (Avg Acc: {best_acc:.4f})")

    print(f"\nFine-tuning complete. Best avg accuracy: {best_acc:.4f}")


if __name__ == "__main__":
    fine_tune_classifier()
```

```python
import os
import torch
import torch.nn.functional as F
import numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from modified_dualresnet import DualResNet


def make_dataloaders(root_dir, image_size=224, batch_size=16, num_workers=2):
    transform = transforms.Compose([
        transforms.Resize((image_size, image_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

    ds_sat = datasets.ImageFolder(os.path.join(root_dir, "gallery_satellite"), transform)
    ds_dr = datasets.ImageFolder(os.path.join(root_dir, "query_drone"), transform)

    return {
        'satellite': DataLoader(ds_sat, batch_size=batch_size, shuffle=False, num_workers=num_workers),
        'drone': DataLoader(ds_dr, batch_size=batch_size, shuffle=False, num_workers=num_workers),
    }, {
        'satellite': ds_sat,
        'drone': ds_dr
    }

import numpy as np
import torch

def compute_mAP(index, good_index, junk_index):
    cmc = torch.IntTensor(len(index)).zero_()
    if len(good_index) == 0:
        return 0.0, cmc

    mask = ~np.isin(index, junk_index)
    index = index[mask]

    order = np.where(np.isin(index, good_index))[0]

    if len(order) == 0:
        return 0.0, cmc

    cmc[order[0]:] = 1

    num_good = len(good_index)
    precision_at_i = [(i + 1) / (rank + 1) for i, rank in enumerate(order)]
    ap = np.sum(precision_at_i) / num_good

    return ap, cmc

def evaluate(qf, ql, gf, gl):
    qf = qf.view(1, -1)
    scores = torch.nn.functional.cosine_similarity(gf, qf, dim=1).cpu().numpy()
    index = np.argsort(scores)[::-1]
    gl = np.asarray(gl)

    good_index = np.argwhere(gl == ql).flatten()
    junk_index = np.argwhere(gl == -1).flatten()

    ap, cmc = compute_mAP(index, good_index, junk_index)
    return ap, cmc


def extract_features(model, loader, view=1):
    model.eval()
    feats, labels = [], []
    with torch.no_grad():
        for imgs, labs in loader:
            if view == 1:
                feat, _, _, _ = model(imgs, torch.zeros_like(imgs))
            else:
                _, feat, _, _ = model(torch.zeros_like(imgs), imgs)
            feat = F.normalize(feat, dim=1)
            feats.append(feat.cpu())
            labels.extend(labs.numpy())
    return torch.cat(feats), np.array(labels)


def test():
    root_dir = '/content/test_dataset_subset'
    weights_path = '/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth'

    dataloaders, datasets = make_dataloaders(root_dir)
    model = DualResNet(num_classes=100, dropout=0.3)
    model.load_state_dict(torch.load(weights_path, map_location='cpu'))
    model = model.to('cpu')
    model.eval()

    #gf, gl = extract_features(model, dataloaders['drone'], view=2)
    #qf, ql = extract_features(model, dataloaders['satellite'], view=1)

    gf, gl = extract_features(model, dataloaders['satellite'], view=1)  # gallery
    qf, ql = extract_features(model, dataloaders['drone'], view=2)


    CMCs, APs = [], []
    for i in range(len(ql)):
        ap, cmc = evaluate(qf[i], ql[i], gf, gl)
        if cmc[0] == -1:
            continue
        CMCs.append(cmc.unsqueeze(0))
        APs.append(ap)

    CMC = torch.cat(CMCs).float().mean(0).numpy()
    mAP = np.mean(APs)
    print(f"Recall@1: {CMC[0]*100:.2f}%, Recall@5: {CMC[4]*100:.2f}%, mAP: {mAP*100:.2f}%")


if __name__ == '__main__':
    test()
```

# Integrating Fusion

```python
import os
import torch
import torch.nn.functional as F
import numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from modified_dualresnet import DualResNet
from collections import defaultdict

def make_dataloaders(root_dir, image_size=224, batch_size=16, num_workers=2):
    transform = transforms.Compose([
        transforms.Resize((image_size, image_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

    ds_sat = datasets.ImageFolder(os.path.join(root_dir, "gallery_satellite"), transform)
    ds_dr = datasets.ImageFolder(os.path.join(root_dir, "query_drone"), transform)

    return {
        'satellite': DataLoader(ds_sat, batch_size=batch_size, shuffle=False, num_workers=num_workers),
        'drone': DataLoader(ds_dr, batch_size=batch_size, shuffle=False, num_workers=num_workers),
    }, {
        'satellite': ds_sat,
        'drone': ds_dr
    }

def compute_mAP(index, good_index, junk_index):
    cmc = torch.IntTensor(len(index)).zero_()
    if len(good_index) == 0:
        return 0.0, cmc

    mask = ~np.isin(index, junk_index)
    index = index[mask]
    order = np.where(np.isin(index, good_index))[0]

    if len(order) == 0:
        return 0.0, cmc

    cmc[order[0]:] = 1
    precision_at_i = [(i + 1) / (rank + 1) for i, rank in enumerate(order)]
    ap = np.sum(precision_at_i) / len(good_index)
    return ap, cmc

def evaluate(qf, ql, gf, gl):
    qf = qf.view(1, -1)
    scores = F.cosine_similarity(gf, qf, dim=1).cpu().numpy()
    index = np.argsort(scores)[::-1]
    gl = np.asarray(gl)
    good_index = np.argwhere(gl == ql).flatten()
    junk_index = np.argwhere(gl == -1).flatten()
    return compute_mAP(index, good_index, junk_index)

def extract_features(model, loader, view=1):
    model.eval()
    feats, labels = [], []
    with torch.no_grad():
        for imgs, labs in loader:
            imgs = imgs.to('cpu')
            if view == 1:
                feat, _, _, _ = model(imgs, torch.zeros_like(imgs))
            else:
                _, feat, _, _ = model(torch.zeros_like(imgs), imgs)
            feat = F.normalize(feat, dim=1)
            feats.append(feat.cpu())
            labels.extend(labs.numpy())
    return torch.cat(feats), np.array(labels)

def multi_query_fusion(qf, ql):
    from collections import defaultdict
    qf = qf.numpy() if isinstance(qf, torch.Tensor) else qf
    label_to_indices = defaultdict(list)
    for idx, label in enumerate(ql):
        label_to_indices[label].append(idx)

    fused_feats, fused_labels = [], []
    for label, indices in label_to_indices.items():
        avg_feat = np.mean(qf[indices], axis=0)
        fused_feats.append(avg_feat)
        fused_labels.append(label)
    return torch.tensor(np.vstack(fused_feats)), np.array(fused_labels)

def test():
    root_dir = '/content/test_dataset_subset'
    weights_path = '/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth'

    dataloaders, datasets = make_dataloaders(root_dir)
    model = DualResNet(num_classes=100, dropout=0.3)
    model.load_state_dict(torch.load(weights_path, map_location='cpu'))
    model = model.to('cpu')
    model.eval()

    # Extract gallery (satellite) features
    gf, gl = extract_features(model, dataloaders['satellite'], view=1)

    # Extract raw query (drone) features and labels
    qf_raw, ql_raw = extract_features(model, dataloaders['drone'], view=2)

    # Apply multi-query fusion
    qf, ql = multi_query_fusion(qf_raw, ql_raw)

    # Evaluate
    CMCs, APs = [], []
    for i in range(len(ql)):
        ap, cmc = evaluate(qf[i], ql[i], gf, gl)
        if cmc[0] == -1:
            continue
        CMCs.append(cmc.unsqueeze(0))
        APs.append(ap)

    CMC = torch.cat(CMCs).float().mean(0).numpy()
    mAP = np.mean(APs)
    print(f"Recall@1: {CMC[0]*100:.2f}%, Recall@5: {CMC[4]*100:.2f}%, mAP: {mAP*100:.2f}%")

if __name__ == '__main__':
    test()
```

```python
import os
import torch
import torch.nn.functional as F
from torchvision import transforms, datasets
from torch.utils.data import DataLoader
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
from modified_dualresnet import DualResNet
from collections import defaultdict

def make_dataloaders(root_dir, image_size=224, batch_size=16, num_workers=2):
    transform = transforms.Compose([
        transforms.Resize((image_size, image_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

    ds_sat = datasets.ImageFolder(os.path.join(root_dir, "gallery_satellite"), transform)
    ds_dr = datasets.ImageFolder(os.path.join(root_dir, "query_drone"), transform)

    return {
        'satellite': DataLoader(ds_sat, batch_size=batch_size, shuffle=False, num_workers=num_workers),
        'drone': DataLoader(ds_dr, batch_size=batch_size, shuffle=False, num_workers=num_workers),
    }, {
        'satellite': ds_sat,
        'drone': ds_dr
    }

def extract_features(model, loader, view=1):
    model.eval()
    feats, labels, paths = [], [], []
    with torch.no_grad():
        for imgs, labs in loader:
            imgs = imgs.to(next(model.parameters()).device)
            if view == 1:
                feat, _, _, _ = model(imgs, torch.zeros_like(imgs))
            else:
                _, feat, _, _ = model(torch.zeros_like(imgs), imgs)
            feat = F.normalize(feat, dim=1)
            feats.append(feat.cpu())
            labels.extend(labs.numpy())
    return torch.cat(feats), labels

def multi_query_fusion(qf, ql, paths):
    qf = qf.numpy()
    label_to_feats = defaultdict(list)
    label_to_paths = defaultdict(list)

    for feat, label, path in zip(qf, ql, paths):
        label_to_feats[label].append(feat)
        label_to_paths[label].append(path)

    fused_feats, fused_labels, rep_paths = [], [], []
    for label in sorted(label_to_feats.keys()):
        avg_feat = np.mean(label_to_feats[label], axis=0)
        fused_feats.append(avg_feat)
        fused_labels.append(label)
        rep_paths.append(label_to_paths[label][0])  # Use one representative image for that class

    return torch.tensor(fused_feats), fused_labels, rep_paths

def visualize_fused_top5(root_dir, weights_path, device='cpu', max_classes=10, save_dir='fused_visualizations'):
    os.makedirs(save_dir, exist_ok=True)

    dataloaders, datasets = make_dataloaders(root_dir, batch_size=1, num_workers=0)

    model = DualResNet(num_classes=100, dropout=0.3)
    model.load_state_dict(torch.load(weights_path, map_location=device))
    model = model.to(device)
    model.eval()

    # Extract features
    gf, gl = extract_features(model, dataloaders['satellite'], view=1)
    qf, ql = extract_features(model, dataloaders['drone'], view=2)

    # Image paths
    drone_img_paths = [path[0] for path in datasets['drone'].imgs]
    sat_img_paths = [path[0] for path in datasets['satellite'].imgs]

    # Apply fusion
    fused_qf, fused_ql, rep_paths = multi_query_fusion(qf, ql, drone_img_paths)

    # Normalize
    gf = F.normalize(gf, dim=1)
    fused_qf = F.normalize(fused_qf, dim=1)

    vis_transform = transforms.Resize((224, 224))

    for i in range(min(max_classes, len(rep_paths))):
        query_feat = fused_qf[i].unsqueeze(0)
        sims = torch.matmul(query_feat, gf.T).squeeze(0)
        top5_idx = sims.topk(5).indices.cpu().numpy()

        # Load and transform images
        drone_img = vis_transform(Image.open(rep_paths[i]).convert('RGB'))
        sat_imgs = [vis_transform(Image.open(sat_img_paths[idx]).convert('RGB')) for idx in top5_idx]

        # Plot side-by-side
        fig, axes = plt.subplots(1, 6, figsize=(18, 4))
        axes[0].imshow(drone_img)
        axes[0].set_title(f"Drone Class {fused_ql[i]}", fontsize=10)
        axes[0].axis('off')

        for j, sat_img in enumerate(sat_imgs):
            axes[j + 1].imshow(sat_img)
            axes[j + 1].set_title(f"Top {j+1}", fontsize=10)
            axes[j + 1].axis('off')

        plt.tight_layout()
        save_path = os.path.join(save_dir, f'fused_drone_class_{fused_ql[i]}_top5.png')
        plt.savefig(save_path)
        plt.show()
        plt.close(fig)
        print(f"Saved and displayed: {save_path}")

if __name__ == "__main__":
    visualize_fused_top5(
        root_dir='/content/test_dataset_subset',
        weights_path='/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth',
        device='cuda' if torch.cuda.is_available() else 'cpu',
        max_classes=10,
        save_dir='fused_visualizations'
    )
```

# Re-ranking

```python
import numpy as np
import torch

def re_ranking(probFea, galFea, k1=20, k2=6, lambda_value=0.3, local_distmat=None, use_gpu=False):
    query_num = probFea.size(0)
    all_num = query_num + galFea.size(0)
    feat = torch.cat([probFea, galFea], dim=0)
    distmat = torch.pow(feat, 2).sum(dim=1, keepdim=True).expand(all_num, all_num)
    distmat = distmat + distmat.t()
    distmat.addmm_(feat, feat.t(), beta=1, alpha=-2)
    original_dist = distmat.cpu().numpy()
    del feat, distmat

    if local_distmat is not None:
        original_dist = original_dist + local_distmat

    gallery_num = original_dist.shape[0]
    original_dist = original_dist / np.max(original_dist, axis=0)
    original_dist = original_dist.T

    V = np.zeros_like(original_dist, dtype=np.float32)
    initial_rank = np.argsort(original_dist).astype(np.int32)

    for i in range(all_num):
        forward_k = initial_rank[i, :k1 + 1]
        backward_k = initial_rank[forward_k, :k1 + 1]
        fi = np.where(backward_k == i)[0]
        k_recip = forward_k[fi]
        k_recip_exp = k_recip.copy()
        for cand in k_recip:
            forward_k2 = initial_rank[cand, :int(np.around(k1 / 2)) + 1]
            backward_k2 = initial_rank[forward_k2, :int(np.around(k1 / 2)) + 1]
            fi2 = np.where(backward_k2 == cand)[0]
            cand_recip = forward_k2[fi2]
            if len(np.intersect1d(cand_recip, k_recip)) > 2/3 * len(cand_recip):
                k_recip_exp = np.append(k_recip_exp, cand_recip)
        k_recip_exp = np.unique(k_recip_exp)
        weight = np.exp(-original_dist[i, k_recip_exp])
        V[i, k_recip_exp] = weight / np.sum(weight)

    original_dist = original_dist[:query_num, :]
    if k2 != 1:
        V_qe = np.zeros_like(V, dtype=np.float32)
        for i in range(all_num):
            V_qe[i, :] = np.mean(V[initial_rank[i, :k2], :], axis=0)
        V = V_qe

    invIndex = [np.where(V[:, i] != 0)[0] for i in range(gallery_num)]
    jaccard_dist = np.zeros_like(original_dist, dtype=np.float32)

    for i in range(query_num):
        temp_min = np.zeros((1, gallery_num), dtype=np.float32)
        non_zero = np.where(V[i, :] != 0)[0]
        for j in non_zero:
            temp_min[0, invIndex[j]] += np.minimum(V[i, j], V[invIndex[j], j])
        jaccard_dist[i] = 1 - temp_min / (2 - temp_min)

    final_dist = jaccard_dist * (1 - lambda_value) + original_dist * lambda_value
    return final_dist
```

# Whitening

```python
import os
import torch
import torch.nn.functional as F
import numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from modified_dualresnet import DualResNet


def make_dataloaders(root_dir, image_size=224, batch_size=16, num_workers=2):
    transform = transforms.Compose([
        transforms.Resize((image_size, image_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

    ds_sat = datasets.ImageFolder(os.path.join(root_dir, "gallery_satellite"), transform)
    ds_dr = datasets.ImageFolder(os.path.join(root_dir, "query_drone"), transform)

    return {
        'satellite': DataLoader(ds_sat, batch_size=batch_size, shuffle=False, num_workers=num_workers),
        'drone': DataLoader(ds_dr, batch_size=batch_size, shuffle=False, num_workers=num_workers),
    }, {
        'satellite': ds_sat,
        'drone': ds_dr
    }


def compute_mAP(index, good_index, junk_index):
    cmc = torch.IntTensor(len(index)).zero_()
    if len(good_index) == 0:
        return 0.0, cmc

    mask = ~np.isin(index, junk_index)
    index = index[mask]

    order = np.where(np.isin(index, good_index))[0]

    if len(order) == 0:
        return 0.0, cmc

    cmc[order[0]:] = 1

    num_good = len(good_index)
    precision_at_i = [(i + 1) / (rank + 1) for i, rank in enumerate(order)]
    ap = np.sum(precision_at_i) / num_good

    return ap, cmc


def evaluate(qf, ql, gf, gl, metric='cosine'):
    qf = qf.view(1, -1)

    if metric == 'cosine':
        scores = F.cosine_similarity(gf, qf, dim=1).cpu().numpy()
    elif metric == 'euclidean':
        dists = torch.cdist(qf, gf, p=2).squeeze(0)  # Euclidean distance
        scores = -dists.cpu().numpy()
    elif metric == 'manhattan':
        dists = torch.cdist(qf, gf, p=1).squeeze(0)  # Manhattan distance
        scores = -dists.cpu().numpy()
    else:
        raise ValueError(f"Unknown metric: {metric}")

    index = np.argsort(scores)[::-1]
    gl = np.asarray(gl)

    good_index = np.argwhere(gl == ql).flatten()
    junk_index = np.argwhere(gl == -1).flatten()

    ap, cmc = compute_mAP(index, good_index, junk_index)
    return ap, cmc


def extract_features(model, loader, view=1):
    model.eval()
    feats, labels = [], []
    with torch.no_grad():
        for imgs, labs in loader:
            imgs = imgs.to(next(model.parameters()).device)
            if view == 1:
                feat, _, _, _ = model(imgs, torch.zeros_like(imgs))
            else:
                _, feat, _, _ = model(torch.zeros_like(imgs), imgs)
            feat = F.normalize(feat, dim=1)
            feats.append(feat.cpu())
            labels.extend(labs.numpy())
    return torch.cat(feats), np.array(labels)


def compute_whitening_params(features: np.ndarray):
    mean = np.mean(features, axis=0)
    cov = np.cov(features.T)
    U, S, _ = np.linalg.svd(cov)
    projection = U @ np.diag(1.0 / np.sqrt(S + 1e-5))
    return mean, projection


class Whitening:
    def __init__(self, mean: np.ndarray, projection: np.ndarray):
        self.mean = mean
        self.projection = projection

    def __call__(self, features: np.ndarray):
        features = features - self.mean
        whitened = np.dot(features, self.projection)
        return whitened


def test():
    root_dir = '/content/test_dataset_subset'
    weights_path = '/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth'

    dataloaders, datasets = make_dataloaders(root_dir)
    model = DualResNet(num_classes=100, dropout=0.3)
    model.load_state_dict(torch.load(weights_path, map_location='cpu'))
    model = model.to('cpu')
    model.eval()

    # Extract gallery and query features
    gf, gl = extract_features(model, dataloaders['satellite'], view=1)
    qf, ql = extract_features(model, dataloaders['drone'], view=2)

    # Compute whitening parameters on gallery features (numpy)
    mean, projection = compute_whitening_params(gf.numpy())
    whiten = Whitening(mean, projection)

    # Apply whitening + normalization
    gf_whiten = F.normalize(torch.from_numpy(whiten(gf.numpy())).float(), dim=1)
    qf_whiten = F.normalize(torch.from_numpy(whiten(qf.numpy())).float(), dim=1)

    # Evaluate with whitening and Euclidean distance (you can change metric)
    metric = 'euclidean'  # or 'cosine', 'manhattan'

    CMCs, APs = [], []
    for i in range(len(ql)):
        ap, cmc = evaluate(qf_whiten[i], ql[i], gf_whiten, gl, metric=metric)
        if cmc[0] == -1:
            continue
        CMCs.append(cmc.unsqueeze(0))
        APs.append(ap)

    CMC = torch.cat(CMCs).float().mean(0).numpy()
    mAP = np.mean(APs)
    print(f"Metric: {metric}")
    print(f"Recall@1: {CMC[0]*100:.2f}%, Recall@5: {CMC[4]*100:.2f}%, mAP: {mAP*100:.2f}%")


if __name__ == '__main__':
    test()
```

```python
import os
import torch
import torch.nn.functional as F
import numpy as np
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from modified_dualresnet import DualResNet


def make_dataloaders(root_dir, image_size=224, batch_size=16, num_workers=2):
    transform = transforms.Compose([
        transforms.Resize((image_size, image_size)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    ds_sat = datasets.ImageFolder(os.path.join(root_dir, "gallery_satellite"), transform)
    ds_dr = datasets.ImageFolder(os.path.join(root_dir, "query_drone"), transform)
    return {
        'satellite': DataLoader(ds_sat, batch_size=batch_size, shuffle=False, num_workers=num_workers),
        'drone': DataLoader(ds_dr, batch_size=batch_size, shuffle=False, num_workers=num_workers),
    }, {
        'satellite': ds_sat,
        'drone': ds_dr
    }


def compute_mAP(index, good_index, junk_index):
    cmc = torch.IntTensor(len(index)).zero_()
    if len(good_index) == 0:
        return 0.0, cmc
    mask = ~np.isin(index, junk_index)
    index = index[mask]
    order = np.where(np.isin(index, good_index))[0]
    if len(order) == 0:
        return 0.0, cmc
    cmc[order[0]:] = 1
    precision_at_i = [(i + 1) / (rank + 1) for i, rank in enumerate(order)]
    ap = np.sum(precision_at_i) / len(good_index)
    return ap, cmc


def evaluate(qf, ql, gf, gl):
    qf = qf.view(1, -1)
    scores = F.cosine_similarity(gf, qf, dim=1).cpu().numpy()
    index = np.argsort(scores)[::-1]
    gl = np.asarray(gl)
    good_index = np.argwhere(gl == ql).flatten()
    junk_index = np.argwhere(gl == -1).flatten()
    ap, cmc = compute_mAP(index, good_index, junk_index)
    return ap, cmc


def extract_features(model, loader, view=1, use_tta=True):
    model.eval()
    feats, labels = [], []
    with torch.no_grad():
        for imgs, labs in loader:
            imgs = imgs.to(next(model.parameters()).device)
            img_zero = torch.zeros_like(imgs)

            if view == 1:
                x, y = imgs, img_zero
            else:
                x, y = img_zero, imgs

            feat_list = []

            # Original
            feat = model(x, y)[0 if view == 1 else 1]
            feat_list.append(feat)

            # TTA: Horizontal flip
            if use_tta:
                x_flip = torch.flip(x, dims=[3])
                y_flip = torch.flip(y, dims=[3])
                feat_flip = model(x_flip, y_flip)[0 if view == 1 else 1]
                feat_list.append(feat_flip)

            # Average features and normalize
            feat = torch.stack(feat_list).mean(dim=0)
            feat = F.normalize(feat, dim=1)

            feats.append(feat.cpu())
            labels.extend(labs.numpy())

    return torch.cat(feats), torch.tensor(labels)


def test():
    root_dir = '/content/test_dataset_subset'  # Update if needed
    weights_path = '/content/ACMMM23-Solution-MBEG/weights/finetuned_best.pth'

    dataloaders, datasets = make_dataloaders(root_dir)
    model = DualResNet(num_classes=100, dropout=0.3)
    state_dict = torch.load(weights_path, map_location='cpu')
    model.load_state_dict(state_dict, strict=False)
    model = model.to('cpu')
    model.eval()

    print("Extracting features with TTA...")

    # Satellite (gallery)
    feat1_sat, labels_sat = extract_features(model, dataloaders['satellite'], view=1, use_tta=True)
    feat2_sat, _ = extract_features(model, dataloaders['satellite'], view=2, use_tta=True)
    gf = torch.cat([feat1_sat, feat2_sat], dim=1)

    # Drone (query)
    feat1_dr, labels_dr = extract_features(model, dataloaders['drone'], view=1, use_tta=True)
    feat2_dr, _ = extract_features(model, dataloaders['drone'], view=2, use_tta=True)
    qf = torch.cat([feat1_dr, feat2_dr], dim=1)

    print("Evaluating retrieval...")
    CMCs, APs = [], []
    for i in range(len(labels_dr)):
        ap, cmc = evaluate(qf[i], labels_dr[i], gf, labels_sat)
        if cmc[0] == -1:
            continue
        CMCs.append(cmc.unsqueeze(0))
        APs.append(ap)

    CMC = torch.cat(CMCs).float().mean(0).numpy()
    mAP = np.mean(APs)

    print(f"\nFinal Results with TTA and Fusion:")
    print(f"Recall@1: {CMC[0]*100:.2f}%, Recall@5: {CMC[4]*100:.2f}%, mAP: {mAP*100:.2f}%")


if __name__ == '__main__':
    test()
```
